```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Flatten
```

```python
print(type(keras.datasets.mnist.load_data()))
```

```python
(X_train, y_train),(X_test, y_test) = keras.datasets.mnist.load_data()
```

```python
X_train.shape
```

```python
X_train[0]
```

```python
y_train
```

```python
import matplotlib.pyplot as plt
plt.imshow(X_train[0]) # image show of grid values
```

```python
X_train = X_train/255
X_test = X_test/255
# all grid values between 0 to 1
```

```python
X_train[0]
```

```python
model = Sequential()

model.add(Flatten(input_shape=(28,28)))
model.add(Dense(128, activation='relu')) # no need to mention
# input_dim as output of flatten will manage it
model.add(Dense(32, activation='relu'))
model.add(Dense(10, activation='softmax'))
```

```python
model.summary()
```

```python
model.compile(optimizer = 'Adam',loss='sparse_categorical_crossentropy', metrics = ['accuracy'])
```

```python
model.fit(X_train, y_train, epochs=25, validation_split=0.2)
```

```python
history = model.fit(X_train, y_train, epochs=25, validation_split=0.2)
```

```python
y_prob = model.predict(X_test) # of every image calculates
# prob of 1, 2, 3 ets.
```

```python
y_prob
```

```python
y_prob.argmax(axis=1)
```

```python
y_pred = y_prob.argmax(axis=1)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test, y_pred)
```

```python
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
```

```python
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
```
