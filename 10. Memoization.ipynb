```python
# fib
import time
def fib(n):
      if n == 1 or n == 0:
            return 1
      else:
            return fib(n-1) + fib(n-2)
```

```python
start = time.time()
fib(36)
print(time.time() - start)
```

```python
# memoization
# fib
def fib_mem(n,d):
      if n in d:
            return d[n]
      else:
            d[n] = fib_mem(n-1,d) + fib_mem(n-2,d)
            return d[n]
```

```python
start = time.time()
d = {0:1, 1:1}
fib_mem(100,d)
print(time.time() - start)
```
